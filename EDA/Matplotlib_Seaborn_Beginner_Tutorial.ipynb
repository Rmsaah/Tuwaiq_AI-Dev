````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
````

## Matplotlib Basics

### Line Plot

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
#plt.grid(False)
plt.show()
````

###Bar Plot

````python
labels = ['A', 'B', 'C']
values = [10, 25, 7]

plt.bar(labels, values)
plt.title("Bar Chart Example")
plt.show()
````

###Pie Chart

````python
sizes = [40, 30, 20, 10]
labels = ['Apple', 'Banana', 'Orange', 'Mango']

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Fruit Distribution")
plt.show()
````

## Seaborn Basics
We will use the built-in `tips` dataset.

### About the `tips` Dataset

We will use the built-in **`tips`** dataset provided by Seaborn.

This dataset contains information about restaurant bills and the corresponding tips given by customers. It is commonly used for learning data visualization and statistical analysis.

####The main columns in the dataset:

| Column       | Description                                 |
| ------------ | ------------------------------------------- |
| `total_bill` | Total amount of the bill                    |
| `tip`        | Tip amount given                            |
| `sex`        | Gender of the customer (Male / Female)      |
| `smoker`     | Whether the customer is a smoker (Yes / No) |
| `day`        | Day of the week (Thur, Fri, Sat, Sun)       |
| `time`       | Time of day (Lunch / Dinner)                |
| `size`       | Number of people in the dining party        |


````python
tips = sns.load_dataset("tips")
tips.head()
````

### Histogram

````python
sns.histplot(tips['total_bill'], kde=True)
plt.title("Histogram of Total Bills")
plt.show()
````

* Most bills are between **10 and 20 dollars** — this means most customers paid that amount.
* Very few customers had bills over **40 dollars**.
* This kind of plot helps us understand what is the most common total bill amount.

In summary: **It helps us see what amounts are most frequent in the data.**




###Box Plot

````python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("Boxplot of Total Bill by Day")
plt.show()
````


There are more outliers on weekends, indicating some customers paid very large bills.

This visualization helps us compare spending patterns across the week.

In summary:
Boxplots let us quickly compare how total bills vary by day and spot unusual values (outliers).

###Scatter Plot

````python
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.title("Total Bill vs Tip")
plt.show()
````

There is a clear positive relationship: as the bill increases, the tip tends to increase too.

Male and female customers show similar tipping behavior.

Some customers gave very high tips compared to their bill – these are seen as points that are higher than expected.

Summary:
This type of plot helps us understand the relationship between two numerical variables, and adding color by category (like gender) gives extra context.

###Heatmap

````python
correlation = tips.select_dtypes(include='number').corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
````


> When using `.corr()` to calculate correlation, make sure to apply it only to **numeric columns**.
> If the dataset contains text (like `'sex'`, `'smoker'`, `'day'`), Python will raise an error because it **cannot convert strings to numbers**.
>
>  Use `.select_dtypes(include='number')` to filter only numeric columns before calculating correlation:
>
> ```python
> correlation = tips.select_dtypes(include='number').corr()
> ```
