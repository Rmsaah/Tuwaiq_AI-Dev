# 1. Understanding Error Types

## Syntax Error

Python cannot understand the code structure.

```python
if score >= 60
    print("Passed")
```

The missing colon prevents Python from parsing the program.

## Runtime Exception

The code is valid, but an error occurs during execution.

```python
result = 10 / 0
```

This raises `ZeroDivisionError`.

## Logical Error

The code runs, but the result is wrong.

```python
area = 2 * (length + width)
```

This formula calculates perimeter, not area.

`try-except` handles runtime exceptions. It does not automatically fix syntax errors or logical mistakes.

## Common Built-in Exceptions

| Exception | Common cause |
|---|---|
| `ValueError` | Invalid value, such as `int("hello")` |
| `TypeError` | Incompatible type or operation |
| `ZeroDivisionError` | Division by zero |
| `IndexError` | Invalid list position |
| `KeyError` | Missing dictionary key |
| `FileNotFoundError` | File does not exist |
| `AttributeError` | Object does not contain the requested attribute |
| `NameError` | Variable name does not exist |

# 2. Basic `try-except`

```python
try:
    risky_operation()
except SomeException:
    recovery_action()
```

Execution flow:

1. Python runs the `try` block.
2. If no exception occurs, the `except` block is skipped.
3. If the matching exception occurs, the remaining statements inside `try` are skipped.
4. Python runs the matching `except` block.
5. The program continues after the structure.

````python
text = "hello"

try:
    number = int(text)
    print("Converted:", number)
except ValueError:
    print("The value is not a valid integer.")

print("Program continues.")
````

# 4. Multiple `except` Blocks

One operation may fail in different ways.

```python
try:
    result = float(first) / float(second)
except ValueError:
    print("Input must be numeric.")
except ZeroDivisionError:
    print("The denominator cannot be zero.")
```

Python checks the handlers from top to bottom and runs the first matching one.

````python
pairs = [("100", "4"), ("ten", "2"), ("8", "0")]

for first, second in pairs:
    try:
        result = float(first) / float(second)
    except ValueError:
        print("Invalid numeric input")
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print(result)
````

# 8. The `finally` Block

`finally` runs whether the operation succeeds or fails.

```python
try:
    ...
except SomeException:
    ...
finally:
    print("Operation complete")
```

Typical uses:

- releasing resources,
- restoring temporary state,
- recording completion,
- cleanup actions.

For files, a `with` statement is usually safer than manually closing the file.

````python
def divide(value):
    try:
        result = 100 / value
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("Operation finished")


divide(4)
print("---")
divide(0)
````

# 10. Re-Raising and Exception Chaining

## Re-Raising

```python
except ValueError:
    print("Conversion failed.")
    raise
```

This logs context, then allows the caller to handle the error.

## Exception Chaining

```python
except ValueError as error:
    raise ValueError(
        "Employee ID must be an integer."
    ) from error
```

`from error` preserves the original cause while adding a clearer application-level message.

# 11. Custom Exceptions

Create a custom exception when the failure has an important business meaning.

```python
class CourseFullError(Exception):
    pass
```

Custom exceptions make targeted handling easier and improve readability.

````python
class CourseFullError(Exception):
    pass


def enroll(current_students, capacity):
    if current_students >= capacity:
        raise CourseFullError("The course is full.")

    return current_students + 1


try:
    enroll(30, 30)
except CourseFullError as error:
    print(error)
````

# 15. Mini Project — Safe Student Record Processor

Process:

```python
records = [
    {"name": "Sara", "score": "95"},
    {"name": "", "score": "80"},
    {"name": "Ali", "score": "hello"},
    {"name": "Nora", "score": "105"},
    {"name": "Omar", "score": "72"},
]
```

## Requirements

Create:

1. `validate_name(name)`
2. `validate_score(score)`
3. `process_student(record)`
4. a list of valid students,
5. an error report for invalid students.

## Rules

- Name cannot be empty.
- Score must be numeric.
- Score must be from 0 to 100.
- Valid records include cleaned name, numeric score, and Passed/Failed status.
- Invalid records include original record, exception type, and message.

## Expected Summary

```text
Valid students: 2
Invalid students: 3
```

````python
records = [
    {"name": "Sara", "score": "95"},
    {"name": "", "score": "80"},
    {"name": "Ali", "score": "hello"},
    {"name": "Nora", "score": "105"},
    {"name": "Omar", "score": "72"},
]

# TODO: Define validation and processing functions.
def validate_name(name: str):
    if len(name)  <= 0:
        raise ValueError("Name must be at least 1 character long.")

    return name

def validate_score(score: int):
    try:
        score_converted = int(score)
    except ValueError:
        raise ValueError("Invalid score")

    if not 0 <= score_converted <= 100:
        raise ValueError("Score must be between 0 and 100.")

    return score_converted

def validate_student(record: dict):
    try:
        name = validate_name(record["name"])
        score = validate_score(record["score"])
    except ValueError as error:
        raise error
    except TypeError as error:
        raise error

    return {"name": name, "score": score}

# TODO: Process all records.
valid_students = []
invalid_students = []
for record in records:
    try:
        current_student = validate_student(record)
        valid_students.append(current_student)
    except ValueError as error:
        invalid_students.append(record)
        #print(error)
    except TypeError as error:
        invalid_students.append(record)
        #print(error)

print("Valid students:", len(valid_students))
print("Invalid students:", len(invalid_students))

print(valid_students)
print(invalid_students)
````

### Exercise 6: Circle Calculations

<details>
<summary><strong>Hint</strong></summary>

Use `3.14159`.

</details>

````python
class Circle:
    def __init__(self, radius: float):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def circumference(self):
        return 2 * 3.14159 * self.radius

circle = Circle(5)
print(f"Area: {circle.area():.2f}")
print(f"Circumference: {circle.circumference():.2f}")
````

#### Expected output
```text
Balance: 1250
```

<details>
<summary><strong>Hint</strong></summary>

Reject negative amounts and excessive withdrawals.

</details>

````python
class BankAccount:
    def __init__(self, name: str, balance: float):
        self.name = name
        self._balance = balance

    def deposit(self, amount: float):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount

    def withdraw(self, amount: float):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError("Insufficient balance.")
        self._balance -= amount

account = BankAccount("Sara", 1000)
account.deposit(500)
account.withdraw(250)
print("Balance:", account._balance)
````

Abstract classes define required interfaces.

### Exercise 27: Abstract Shape

**Level:** Challenge  
**Learning objective:** Use ABC and abstractmethod.

#### Task
Create Circle and Rectangle with area methods.

#### Expected output
```text
Two area values
```

<details>
<summary><strong>Hint</strong></summary>

Import from `abc`.

</details>

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

shapes = [Circle(5), Rectangle(4, 6)]

for shape in shapes:
    print(round(shape.area(), 2))
````

### Exercise 30: Smart Clinic System

**Level:** Stretch Project  
**Learning objective:** Design a complete OOP application.

#### Task
Build Person, Patient, Doctor, Appointment, and Clinic classes.

#### Expected output
```text
Complete clinic demonstration
```

<details>
<summary><strong>Hint</strong></summary>

Use inheritance, properties, composition, custom exceptions, and special methods.

</details>

````python
# TODO: Design and implement the complete clinic system.
class AppointmentError(Exception): pass

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value


class Patient(Person):
    def __init__(self, name, age, patient_id):
        super().__init__(name, age)
        self.patient_id = patient_id

class Doctor(Person):
    def __init__(self, name, age, specialty):
        super().__init__(name, age)
        self.specialty = specialty

class Appointment:
    def __init__(self, doctor, patient, date):
        self.doctor = doctor
        self.patient = patient
        self.date = date

    def __str__(self):
        return f"Appointment: {self.patient.name} with Dr. {self.doctor.name} on {self.date}"

class Clinic:
    def __init__(self, name):
        self.name = name
        self.appointments = []

    def schedule(self, doctor, patient, date):
        # Simple validation: ensure doctor and patient are correct types
        if not isinstance(doctor, Doctor) or not isinstance(patient, Patient):
            raise AppointmentError("Invalid Doctor or Patient provided")
        new_appt = Appointment(doctor, patient, date)
        self.appointments.append(new_appt)
        return new_appt

clinic = Clinic("HealthFirst")
doctor = Doctor("Smith", 45, "Cardiology")
patient = Patient("Sara", 30, "P001")

try:
    clinic.schedule(doctor, patient, "2026-07-20")
    print(f"Clinic: {clinic.name}")
    for appointment in clinic.appointments:
        print(appointment)
except AppointmentError as e:
    print(f"Error: {e}")
````

# OOP Design Guidance

Before creating a class, ask:

1. What real or conceptual object does it represent?
2. What data belongs to it?
3. What behavior belongs to it?
4. Which values require validation?
5. Should this behavior belong to another class?
6. Is inheritance truly an is-a relationship?
7. Would composition be clearer?

## Common Mistakes

- creating one class that does everything,
- using inheritance only to reuse code,
- exposing internal attributes without validation,
- printing inside every method instead of returning values,
- forgetting to update both sides of a relationship,
- creating classes with no meaningful behavior.

````python
# FINAL INTEGRATION PROJECT WORKSPACE

# TODO: Define custom exceptions.
class TrainingCenterError(Exception): pass
class DuplicateStudentError(TrainingCenterError): pass
class DuplicateCourseError(TrainingCenterError): pass
class StudentNotFoundError(TrainingCenterError): pass
class CourseNotFoundError(TrainingCenterError): pass
class CourseFullError(TrainingCenterError): pass
class InvalidEmailError(TrainingCenterError): pass

# TODO: Define validation functions.
def validate_id(student_id):
    if not isinstance(student_id, int) or student_id <= 0:
        raise ValueError("Student ID must be a positive integer.")
    return student_id

def validate_email(email):
    if "@" not in email or "." not in email.split("@")[-1]:
        raise InvalidEmailError("Email must contain one '@' and a dot after it.")
    return email

# TODO: Define Student.
class Student:
    def __init__(self, student_id, name, email):
        self.student_id = validate_id(student_id)
        if not name:
            raise ValueError("Student name cannot be empty.")
        self.name = name
        self.email = validate_email(email)
        self.courses = set()

    def enroll_in_course(self, course_code):
        self.courses.add(course_code)

    def get_summary(self):
        return f"Student {self.name} is enrolled in {len(self.courses)} courses."

    def __str__(self):
        return f"{self.name} (ID: {self.student_id})"

# TODO: Define Course.
class Course:
    STATUS_ACTIVE = True

    def __init__(self, code, title, capacity):
        if not code:
            raise ValueError("Course code cannot be empty.")
        self.code = code
        self.title = title
        if capacity <= 0:
            raise ValueError("Capacity must be a positive integer.")
        self.capacity = capacity
        self.student_ids = []

    def has_capacity(self):
        return len(self) < self.capacity

    def enroll_student(self, student_id):
        self.student_ids.append(student_id)

    def __len__(self):
        return len(self.student_ids)

    def __str__(self):
        return f"{self.title} [{self.code}] ({len(self)}/{self.capacity})"

# TODO: Define TrainingCenter.
class TrainingCenter:
    def __init__(self, name):
        self.name = name
        self.students = {}
        self.courses = {}

    def add_student(self, student):
        if student.student_id in self.students:
            raise DuplicateStudentError(f"Student {student.student_id} already exists.")
        self.students[student.student_id] = student

    def add_course(self, course):
        if course.code in self.courses:
            raise DuplicateCourseError(f"Course {course.code} already exists.")
        self.courses[course.code] = course

    def find_student(self, student_id):
        if student_id not in self.students:
            raise StudentNotFoundError(f"Student {student_id} not found.")
        return self.students[student_id]

    def find_course(self, course_code):
        if course_code not in self.courses:
            raise CourseNotFoundError(f"Course {course_code} not found.")
        return self.courses[course_code]

    def enroll(self, student_id, course_code):
        student = self.find_student(student_id)
        course = self.find_course(course_code)

        if not course.has_capacity():
            raise CourseFullError(f"Course {course_code} is full.")
        if course_code in student.courses:
            return

        student.enroll_in_course(course_code)
        course.enroll_student(student_id)

    def process_requests(self, requests):
        successes = []
        failures = []
        total_processed = 0

        for sid, ccode in requests:
            try:
                self.enroll(sid, ccode)
            except TrainingCenterError as e:
                failures.append({
                    "request": (sid, ccode),
                    "type": type(e).__name__,
                    "message": str(e)
                })
            else:
                successes.append((sid, ccode))
            finally:
                total_processed += 1

        return successes, failures, total_processed

    def produce_summary(self, successes, failures):
        print("Successful Enrollments:")
        for s in successes:
            print(f"Student {s[0]} -> {s[1]}")

        print("\nFailed Enrollments:")
        for f in failures:
            print(f"Req: {f['request']} | {f['type']}: {f['message']}")

        print(f"\nSummary for {self.name}:")
        print(f"Students: {len(self.students)}")
        print(f"Courses: {len(self.courses)}")
        print(f"Successful enrollments: {len(successes)}")
        print(f"Failed enrollments: {len(failures)}")

# TODO:
# 1. Create a training center.
center = TrainingCenter("Tech Academy")

# 2. Add at least three students.
try:
    center.add_student(Student(101, "Sara", "sara@example.com"))
    center.add_student(Student(102, "Ali", "ali@test.org"))
    center.add_student(Student(103, "Nora", "nora@web.com"))
except (ValueError, TrainingCenterError) as e:
    print(f"Error adding student: {e}")

# 3. Add at least two courses.
try:
    center.add_course(Course("PY101", "Python Basics", 2))
    center.add_course(Course("AI202", "AI Systems", 10))
except (ValueError, TrainingCenterError) as e:
    print(f"Error adding course: {e}")

# 4. Process the enrollment requests.
requests = [
    (101, "PY101"),
    (102, "PY101"),
    (999, "PY101"),
    (103, "MISSING"),
    (103, "PY101"),
]

# 5. Store successful and failed requests.
successes, failures, total = center.process_requests(requests)

# 6. Print detailed results and the final summary.
center.produce_summary(successes, failures)
````
