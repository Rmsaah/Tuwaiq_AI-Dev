# Part 3 — Validation Functions

Create reusable functions for:

1. cleaning a full name,
2. validating an email address,

4. validating a non-empty course code.

## Required Behaviors

### `clean_name(name)`
- remove surrounding spaces,
- replace repeated internal spaces,
- convert to title case.

### `validate_email(email)`
- must contain exactly one `@`,
- must contain a dot after `@`,
- must not contain spaces.

### `validate_score(value)`
- convert the value to `float`,
- accept only values from 0 to 100,
- raise `InvalidScoreError` otherwise.

### `validate_course_code(code)`
- remove spaces,
- convert to uppercase,
- reject an empty value.

````python
def clean_name(name: str) -> str:
    stripped_name = name.strip()
    if len(stripped_name) == 0:
        raise ValueError("Error: Empty name field", name)

    cleaned = " ".join(stripped_name.split())
    title_case = cleaned.title()
    return title_case


def validate_email(email: str) -> str:
    stripped_email = email.strip()

    if len(stripped_email) == 0:
        raise ValueError("Error: Empty email field", email)

    find_index = stripped_email.find("@")
    if find_index == -1 or stripped_email.count("@") != 1:
        raise ValueError("Error: Email must contain exactly one '@' symbol", email)

    flag_dot = False
    dot_counter = 0
    while find_index < len(stripped_email):
        if stripped_email[find_index] == ".":
            flag_dot = True
            dot_counter += 1
        find_index += 1
    if not flag_dot or dot_counter != 1:
        raise ValueError("Error: Email should contain exactly one '.' symbol after the '@' symbol", email)

    if " " in stripped_email:
        raise ValueError("Error: Email should not contain any spaces", email)

    return stripped_email


def validate_score(value: Any) -> float:
    try:
        float_value = float(value)
    except (ValueError, TypeError):
        raise InvalidScoreError("Error: Invalid score value", value)

    if not MIN_SCORE <= float_value <= MAX_SCORE:
        raise InvalidScoreError("Error: Invalid score range", float_value)

    return float_value


def validate_course_code(code: str) -> str:
    cleaned_code = code.strip().upper()
    if len(cleaned_code) == 0:
        raise ValueError("Error: Empty course code", cleaned_code)

    return cleaned_code
````

# Part 4 — Class `Trainee`

Create a class named `Trainee`.

## Required Attributes

- `trainee_id`
- `name`
- `email`
- `enrolled_courses`
- `attendance`
- `scores`

## Recommended Data Structures

- `enrolled_courses`: set of course codes
- `attendance`: dictionary mapping course code to a set of attended session numbers
- `scores`: dictionary mapping course code to a list of scores

## Required Methods

- `enroll(course_code)`
- `record_attendance(course_code, session_number)`
- `add_score(course_code, score)`
- `average_score(course_code)`
- `to_dict()`
- `from_dict()`
- `__str__()`

````python
class Trainee:
    def __init__(
        self,
        trainee_id: int,
        name: str,
        email: str
    ):
        # TODO: Validate and store core attributes.
        # TODO: Create empty data structures.
        self.trainee_id = trainee_id
        self.name = name
        self.email = email
        self.enrolled_courses = set() # {course codes}
        self.attendance = {} # {"course code": {attended session numbers}}
        self.scores = {} # {"course code": [scores]}

    @property
    def trainee_id(self) -> int:
        return self._trainee_id

    @trainee_id.setter
    def trainee_id(self, trainee_id: int) -> None:
        self._trainee_id = int(trainee_id)

    @property
    def name(self) -> str:
        return self._name

    @name.setter
    def name(self, name: str) -> None:
        cleaned_name = clean_name(name)
        self._name = cleaned_name

    @property
    def email(self) -> str:
        return self._email

    @email.setter
    def email(self, email: str) -> None:
        validated_email = validate_email(email)
        self._email = validated_email.lower()

    def enroll(self, course_code: str) -> None:
        validated_course_code = validate_course_code(course_code)
        self.enrolled_courses.add(validated_course_code)

        if validated_course_code not in self.attendance:
            self.attendance[validated_course_code] = set()

        if validated_course_code not in self.scores:
            self.scores[validated_course_code] = []

    def record_attendance(
        self,
        course_code: str,
        session_number: int,
    ) -> None:
        validated_course_code = validate_course_code(course_code)

        # if not enrolled
        self.enroll(validated_course_code)
        self.attendance[validated_course_code].add(session_number)

    def add_score(
        self,
        course_code: str,
        score: float,
    ) -> None:
        validated_course_code = validate_course_code(course_code)
        validated_score = validate_score(score)

        # if not enrolled
        self.enroll(validated_course_code)
        self.scores[validated_course_code].append(validated_score)


    def average_score(self, course_code: str) -> float:
        validated_course_code = validate_course_code(course_code)

        all_scores = sum(self.scores.get(validated_course_code, []))
        count_scores = len(self.scores.get(validated_course_code, []))
        try:
            average = all_scores / count_scores
        except ZeroDivisionError:
            average = 0.0

        return average

    def to_dict(self) -> dict:
        # TODO: Convert sets to JSON-compatible lists.
        trainee_dict = {
            "trainee_id": self.trainee_id,
            "name": self.name,
            "email": self.email,
            "enrolled_courses": list(self.enrolled_courses),
            "attendance": {course: list(sessions) for course, sessions in self.attendance.items()},
            "scores": self.scores
        }
        return trainee_dict

    @classmethod
    def from_dict(cls, data: dict) -> "Trainee":
        # TODO: Restore all trainee data.
        trainee_id = data["trainee_id"]
        name = data["name"]
        email = data["email"]
        restored_trainee = cls(trainee_id, name, email)

        # restoring enrolled courses
        restored_trainee.enrolled_courses = set(data.get("enrolled_courses", []))

        # restoring attendances
        restores_attendances = {}
        for course_code, sessions in data.get("attendance", {}).items():
            restores_attendances[course_code] = set(sessions)
        restored_trainee.attendance = restores_attendances

        # restoring scores
        restored_trainee.scores = dict(data.get("scores", {}))

        return restored_trainee

    def __str__(self) -> str:
        string = f"Trainee: {self.name} #{self.trainee_id}\nEnrolled Courses: {self.enrolled_courses}\nAttendance: {self.attendance}\nScores: {self.scores}"
        return string
````

# Part 5 — Class `Course`

Create a class named `Course`.

## Required Attributes

- `code`
- `title`
- `capacity`
- `passing_score`
- `minimum_attendance`
- `schedule`
- `enrolled_trainees`

Use a tuple for `schedule` and a set for `enrolled_trainees`.

## Required Methods

- `has_space()`
- `enroll_trainee(trainee_id)`
- `is_enrolled(trainee_id)`
- `to_dict()`
- `from_dict()`
- `__str__()`

````python
class Course:
    def __init__(
        self,
        code: str,
        title: str,
        capacity: int,
        passing_score: float = DEFAULT_PASSING_SCORE,
        minimum_attendance: float = DEFAULT_MIN_ATTENDANCE,
        schedule: tuple[str, ...] = (),
    ):
        self.code = code
        self.title = title
        self.capacity = capacity
        self.passing_score = passing_score
        self.minimum_attendance = minimum_attendance
        self.schedule = schedule
        self.enrolled_trainees: set[int] = set() # {trainee ids}

    @property
    def code(self) -> str:
        return self._code

    @code.setter
    def code(self, code: str) -> None:
        validated_course_code = validate_course_code(code)
        self._code = validated_course_code

    @property
    def title(self) -> str:
        return self._title

    @title.setter
    def title(self, title: str) -> None:
        cleaned_title = clean_name(title)
        self._title = cleaned_title

    @property
    def capacity(self) -> int:
        return self._capacity

    @capacity.setter
    def capacity(self, capacity: int) -> None:
        self._capacity = capacity


    def has_space(self) -> bool:
        count_enrolled_trainees = len(self.enrolled_trainees)
        return count_enrolled_trainees < self.capacity


    def enroll_trainee(self, trainee_id: int) -> None:
        # TODO: Raise CourseFullError when appropriate.
        if self.is_enrolled(trainee_id):
            raise DuplicateTraineeError("Error: Trainee already enrolled", trainee_id)

        if not self.has_space():
            raise CourseFullError("Error: Course is Full")

        # note: search by trainee id
        self.enrolled_trainees.add(trainee_id)


    def is_enrolled(self, trainee_id: int) -> bool:
        if trainee_id in self.enrolled_trainees:
            return True
        return False


    def to_dict(self) -> dict:
        course_dict = {
            "code": self.code,
            "title": self.title,
            "capacity": self.capacity,
            "passing_score": self.passing_score,
            "minimum_attendance": self.minimum_attendance,
            "schedule": self.schedule,
            "enrolled_trainees": list(self.enrolled_trainees)
        }
        return course_dict


    @classmethod
    def from_dict(cls, data: dict) -> "Course":
        # TODO
        code = data["code"]
        title = data["title"]
        capacity = data["capacity"]
        passing_score = data["passing_score"]
        minimum_attendance = data["minimum_attendance"]

        # restore schedule. JSON saves tuples as lists
        schedule = tuple(data.get("schedule", ()))

        # restore course object
        restored_course = cls(
            code=code,
            title=title,
            capacity=capacity,
            passing_score=passing_score,
            minimum_attendance=minimum_attendance,
            schedule=schedule
        )

        # restore enrolled trainees
        restored_course.enrolled_trainees = set(data.get("enrolled_trainees", []))

        return restored_course

    def __str__(self) -> str:
        string = f"Course: {self.code} - {self.title}\nCapacity: {self.capacity}\nEnrolled Trainees: {self.enrolled_trainees}\nHas space: {self.has_space()}\nSchedule: {self.schedule}"
        return string
````

# Part 6 — Class `TrainingCenter`

This class controls the complete system.

## Required Attributes

- `name`
- `trainees`: dictionary mapping trainee ID to `Trainee`
- `courses`: dictionary mapping course code to `Course`

## Required Methods

### Trainee and course management
- `add_course(course)`
- `register_trainee(trainee)`
- `get_course(course_code)`
- `get_trainee(trainee_id)`
- `enroll_trainee(trainee_id, course_code)`

### Data recording
- `record_attendance(trainee_id, course_code, session_number)`
- `record_score(trainee_id, course_code, score)`

### Calculations
- `attendance_percentage(trainee_id, course_code)`
- `course_status(trainee_id, course_code)`
- `certificate_eligible(trainee_id, course_code)`

### Reports
- `trainee_report(trainee_id)`
- `course_report(course_code)`
- `overall_summary()`

### Files
- `save_json(path)`
- `load_json(path)`
- `export_course_report_csv(course_code, path)`

````python
class TrainingCenter:
    def __init__(self, name: str):
        self.name = name
        self.trainees: dict[int, Trainee] = {} # {trainee id --> trainee object}
        self.courses: dict[str, Course] = {} # {course code --> course object}

    @property
    def name(self) -> str:
        return self._name

    @name.setter
    def name(self, name: str) -> None:
        cleaned_name = clean_name(name)
        self._name = cleaned_name


    def add_course(self, course: Course) -> None:
        course_code = course.code

        if course_code in self.courses:
            raise ValueError("Error: Course code already exists", course_code)

        self.courses[course_code] = course


    def register_trainee(self, trainee: Trainee) -> None:
        trainee_id = trainee.trainee_id

        if trainee_id in self.trainees:
            raise DuplicateTraineeError("Error: Trainee ID already registered", trainee_id)

        self.trainees[trainee_id] = trainee


    def get_course(self, course_code: str) -> Course:
        code = validate_course_code(course_code)
        if code not in self.courses:
            raise CourseNotFoundError("Error: Course not found", course_code)
        return self.courses[code]

    def get_all_courses(self) -> dict:
        return self.courses


    def get_trainee(self, trainee_id: int) -> Trainee:
        if trainee_id not in self.trainees:
            raise TraineeNotFoundError("Error: Trainee not found", trainee_id)
        return self.trainees[trainee_id]

    def get_all_trainees(self) -> dict:
        return self.trainees


    def enroll_trainee(
        self,
        trainee_id: int,
        course_code: str,
    ) -> None:
        # TODO: Update both Trainee and Course objects.
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        course.enroll_trainee(trainee_id) # capacity + duplicate checks first
        trainee.enroll(course_code)


    def record_attendance(
        self,
        trainee_id: int,
        course_code: str,
        session_number: int,
    ) -> None:
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        if not course.is_enrolled(trainee_id):         # rejects recording before enrolling
            raise ValueError("Error: Trainee not enrolled in this course", (trainee_id, course.code))

        trainee.record_attendance(course.code, session_number)


    def record_score(
        self,
        trainee_id: int,
        course_code: str,
        score: float,
    ) -> None:
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        if not course.is_enrolled(trainee_id):
            raise ValueError("Error: Trainee not enrolled in this course", (trainee_id, course.code))

        trainee.add_score(course.code, score)


    def attendance_percentage(
        self,
        trainee_id: int,
        course_code: str,
    ) -> float:
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        scheduled_sessions = len(course.schedule)
        if scheduled_sessions == 0:
            return 0.0

        attended_sessions = len(trainee.attendance.get(course.code, set()))
        return attended_sessions / scheduled_sessions * 100


    def course_status(
        self,
        trainee_id: int,
        course_code: str,
    ) -> str:
        # TODO: Return Passed, Failed, or Incomplete.
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        # check if trainee is enrolled in course
        if course.code not in trainee.enrolled_courses:
            return "Incomplete"

        course_scores = trainee.scores.get(course.code, [])
        if not course_scores:    # no score recorded --> Incomplete
            return "Incomplete"

        average_score = trainee.average_score(course.code)
        attendance = self.attendance_percentage(trainee_id, course.code)

        if average_score >= course.passing_score and attendance >= course.minimum_attendance:
            return "Passed"
        return "Failed"


    def certificate_eligible(
        self,
        trainee_id: int,
        course_code: str,
    ) -> bool:

        # check if trainee passes course
        course_status = self.course_status(trainee_id, course_code)
        return course_status == "Passed"


    def trainee_report(self, trainee_id: int) -> dict:
        trainee = self.get_trainee(trainee_id)

        courses = {}
        for course_code in sorted(trainee.enrolled_courses):
            courses[course_code] = {
                "average": round(trainee.average_score(course_code), 2),
                "attendance": round(self.attendance_percentage(trainee_id, course_code), 2),
                "status": self.course_status(trainee_id, course_code),
                "certificate_eligible": self.certificate_eligible(trainee_id, course_code),
            }

        return {
            "trainee_id": trainee.trainee_id,
            "name": trainee.name,
            "email": trainee.email,
            "courses": courses,
        }


    def course_report(self, course_code: str) -> list[dict]:
        course = self.get_course(course_code)

        rows = []
        for trainee_id in course.enrolled_trainees:
            trainee = self.get_trainee(trainee_id)
            rows.append({
                "trainee_id": trainee_id,
                "trainee_name": trainee.name,
                "average_score": round(trainee.average_score(course.code), 2),
                "attendance_percentage": round(self.attendance_percentage(trainee_id, course.code), 2),
                "status": self.course_status(trainee_id, course.code),
                "certificate_eligible": self.certificate_eligible(trainee_id, course.code),
            })

        rows.sort(key=lambda row: row["average_score"], reverse=True)
        return rows


    def search_trainees(self, text: str) -> list[Trainee]:
        cleaned_text = clean_name(text)
        return [trainee for trainee in self.trainees.values() if cleaned_text in trainee.name]


    def course_ranking(self) -> list[tuple[str, float]]:
        ranking = []
        for course_code, course in self.courses.items():
            trainee_averages = []
            for trainee_id in course.enrolled_trainees:
                trainee = self.get_trainee(trainee_id)
                if trainee.scores.get(course.code):          # only trainees who have scores
                    trainee_averages.append(trainee.average_score(course.code))

            if trainee_averages:
                ranking.append((course_code, sum(trainee_averages) / len(trainee_averages)))

        ranking.sort(key=lambda row: row[1], reverse=True)
        return ranking


    def overall_summary(self) -> dict:
        passed = failed = incomplete = 0
        total_enrollments = 0
        certificates = 0

        for course_code, course in self.courses.items():
            for trainee_id in course.enrolled_trainees:
                total_enrollments += 1

                status = self.course_status(trainee_id, course_code)
                if status == "Passed":
                    passed += 1
                elif status == "Failed":
                    failed += 1
                else:
                    incomplete += 1

                if self.certificate_eligible(trainee_id, course_code):
                    certificates += 1

        ranking = self.course_ranking()
        top_course = ranking[0][0] if ranking else None

        return {
            "num_courses": len(self.courses),
            "num_trainees": len(self.trainees),
            "total_enrollments": total_enrollments,
            "passed_records": passed,
            "failed_records": failed,
            "incomplete_records": incomplete,
            "top_course": top_course,
            "certificates_eligible": certificates,
        }


    def save_json(self, path: Path) -> None:
        data = {
              "name": self.name,
              "courses": [course.to_dict() for course in self.courses.values()],
              "trainees": [trainee.to_dict() for trainee in self.trainees.values()],
        }

        with open(path, "w", encoding="utf-8") as file:
            json.dump(data, file, indent=4)


    @classmethod
    def load_json(cls, path: Path) -> "TrainingCenter":
        with open(path, "r", encoding="utf-8") as file:
              data = json.load(file)

        center = cls(data["name"])
        for course_data in data.get("courses", []):
            center.add_course(Course.from_dict(course_data))
        for trainee_data in data.get("trainees", []):
            center.register_trainee(Trainee.from_dict(trainee_data))
        return center


    def export_course_report_csv(
        self,
        course_code: str,
        path: Path,
    ) -> None:
        rows = self.course_report(course_code)
        fieldnames = [
              "trainee_id",
              "trainee_name",
              "average_score",
              "attendance_percentage",
              "status",
              "certificate_eligible",
        ]
        with open(path, "w", newline="", encoding="utf-8") as file:
            writer = csv.DictWriter(file, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(rows)
````

# Part 19 — Menu-Driven Application

Create a console menu.

## Required Options

```text
1. List courses
2. List trainees
3. Register trainee
4. Enroll trainee
5. Record attendance
6. Record score
7. Show trainee report
8. Show course report
9. Show overall summary
10. Save data
0. Exit
```

Requirements:

- use a `while` loop,
- validate numeric menu choices,
- use `try-except`,
- call the appropriate methods,
- do not duplicate business logic inside the menu.

For automatic notebook testing, implement the menu in a function but do not call it automatically.

````python
def print_menu() -> None:
    print()
    print("=" * 42)
    print(" 1. List courses")
    print(" 2. List trainees")
    print(" 3. Register trainee")
    print(" 4. Enroll trainee")
    print(" 5. Record attendance")
    print(" 6. Record score")
    print(" 7. Show trainee report")
    print(" 8. Show course report")
    print(" 9. Show overall summary")
    print("10. Save data")
    print("11. Search trainee by name")
    print("12. Course ranking")
    print("13. Command history")
    print(" 0. Exit")
    print("=" * 42)


def ask_int(prompt: str) -> int:
    while True:
        answer = input(prompt).strip()
        try:
            return int(answer)
        except ValueError:
            print("Please enter a whole number.")


def run_menu(center: TrainingCenter) -> None:
    history = []

    while True:
        print_menu()
        choice = ask_int("Enter a number to choose from the menu: ")
        print()

        if choice == 0:
            print("Program closing...")
            break

        history.append(choice)

        try:
            match choice:
                case 1: # list courses
                    for code, course in center.get_all_courses().items():
                        print(f"  {code}  {course.title}  "
                              f"({len(course.enrolled_trainees)}/{course.capacity} enrolled)")

                case 2: # list trainees
                    for trainee_id, trainee in center.get_all_trainees().items():
                        print(f"  #{trainee_id}  {trainee.name}  <{trainee.email}>")

                case 3: # register trainee
                    new_trainee = Trainee(
                        ask_int("Enter the new trainee's ID: "),
                        input("Enter the new trainee's name: "),
                        input("Enter the new trainee's email: "),
                    )
                    center.register_trainee(new_trainee)
                    print(f"Registered #{new_trainee.trainee_id} {new_trainee.name}.")

                case 4: # enroll trainee
                    center.enroll_trainee(
                        ask_int("Enter trainee ID: "),
                        input("Enter course code: "),
                    )
                    print("Trainee enrolled.")

                case 5: # record attendance
                    center.record_attendance(
                        ask_int("Enter trainee ID: "),
                        input("Enter course code: "),
                        ask_int("Enter session number: "),
                    )
                    print("Attendance recorded.")

                case 6: # record score
                    center.record_score(
                        ask_int("Enter trainee ID: "),
                        input("Enter course code: "),
                        input("Enter score: "),
                    )
                    print("Score recorded.")

                case 7: # show trainee report
                    print_trainee_report(center, ask_int("Enter trainee ID: "))

                case 8: # show course report
                    print_course_report(center, input("Enter course code: "))

                case 9: # show overall summary
                    print_overall_summary(center)

                case 10: # save data
                    save_path = Path.cwd() / "training_center.json"
                    center.save_json(save_path)
                    print("Data saved to:", save_path)

                case 11: # search trainee by name
                    found_trainees = center.search_trainees(input("Enter part of a name: "))
                    if not found_trainees:
                        print("  No trainee matched.")
                    for trainee in found_trainees:
                        print(f"  #{trainee.trainee_id}  {trainee.name}  <{trainee.email}>")

                case 12: # course ranking
                    for position, (course_code, average) in enumerate(center.course_ranking(), start=1):
                        print(f"  {position}. {course_code}  "
                              f"{center.get_course(course_code).title}  {average:.2f}")

                case 13: # command history
                    print("Options used so far:", history)

                case _:
                    print("Please enter a number from the menu.")

        except (ValueError, TypeError, DuplicateTraineeError, CourseFullError,
                TraineeNotFoundError, CourseNotFoundError) as error:
            print("Error:", error)


# Do not call run_menu(center) automatically.
# Run it manually after all other cells are complete:   run_menu(center)
````
