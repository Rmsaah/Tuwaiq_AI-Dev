# LLMs for Beginners: Embeddings

## What is an LLM?
An LLM (Large Language Model) learns patterns from a large amount of text. It can generate text, answer questions, summarize documents, translate, and write code.

## What is an embedding?
An embedding is a list of numbers representing the meaning of text. Texts with related meanings tend to have similar vectors. This is useful for search, recommendations, clustering, and finding related documents.

## Main idea

```text
Text -> Embedding vector -> Similarity score
```

An embedding does not generate a paragraph. It represents meaning so that we can compare texts mathematically.

## 1. Setup

````python
!pip install -q -U openai numpy

from getpass import getpass
import numpy as np
from openai import OpenAI

api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
EMBEDDING_MODEL = "text-embedding-3-small"
print("Setup complete")
````

## 2. Text data

````python
query = "I enjoy learning about artificial intelligence."

sentences = [
    "Machine learning helps computers learn from data.",
    "I like studying AI and new technology.",
    "The football match was exciting yesterday.",
    "Pasta is one of my favorite meals.",
]

print(query)
print(sentences)
````

## 3. Create embeddings

````python
def get_embedding(text):
    response = client.embeddings.create(
        model=EMBEDDING_MODEL,
        input=text,
    )
    return response.data[0].embedding

query_embedding = get_embedding(query)
print("Vector length:", len(query_embedding))
````

## 4. Cosine similarity

Cosine similarity compares the direction of two vectors. A larger score generally means that the texts are more related in meaning.

````python
def cosine_similarity(vector_a, vector_b):
    vector_a = np.array(vector_a)
    vector_b = np.array(vector_b)
    denominator = np.linalg.norm(vector_a) * np.linalg.norm(vector_b)
    return float(np.dot(vector_a, vector_b) / denominator)

example_a = np.array([1.0, 0.0])
example_b = np.array([0.8, 0.2])
print("Example similarity:", cosine_similarity(example_a, example_b))
````

## 5. Find the most similar sentence

````python
sentence_embeddings = [get_embedding(sentence) for sentence in sentences]
scores = [cosine_similarity(query_embedding, embedding) for embedding in sentence_embeddings]
best_index = int(np.argmax(scores))

for sentence, score in zip(sentences, scores):
    print(f"{score:.4f} - {sentence}")

print("\nBest match:", sentences[best_index])
print("Best score:", scores[best_index])
````

## What did we learn?

- An LLM generates or transforms language.
- An embedding represents the meaning of text as numbers.
- Cosine similarity lets us compare two embeddings.
- Embeddings can power semantic search: search by meaning instead of exact keywords.
